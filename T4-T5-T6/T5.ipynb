# This is the *T4 & T5* Section

```python
#import libraries needed for T4 and T5
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error,
    explained_variance_score
)
```

```python
#load csv
df = pd.read_csv("diabetes_data.csv")
display(df.head())
print("Dataset shape:", df.shape)
df.info()
```

```python
#clean by checking missing values
print("Missing values:")
print(df.isnull().sum())
display(df.describe())
```

```python
#separate x and y features
X = df.drop("target", axis=1)
y = df["target"]
print("Input variables:")
print(X.columns.tolist())
print("\nTarget variable:")
print(y.name)
```

```python
#split for train/test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
#apply multi-linear regression
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("multi linear regression model trained.")
```

```python
#display b0, b1, ..., bn coefficients
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})
display(coefficients)
print("Intercept:", model.intercept_)
```

# This is the *T4* Section

```python
#calculate performance metrics
r2 = r2_score(y_test, y_pred) #variation in y explained by model
mse = mean_squared_error(y_test, y_pred) #average of squared errors
rmse = np.sqrt(mse) #sqrt of mse
mae = mean_absolute_error(y_test, y_pred) #average of abs errors
print("Model Performance")
print(f"R² Score           : {r2:.4f}")
print(f"MAE                : {mae:.4f}")
print(f"MSE                : {mse:.4f}")
print(f"RMSE               : {rmse:.4f}")
```
